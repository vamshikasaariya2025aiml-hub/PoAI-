```python
#Map Coloring Problem

colors = ['Red', 'Green', 'Blue']

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

color = {}

def safe(node, c):
    for neighbor in graph[node]:
        if color.get(neighbor) == c:
            return False
    return True

def solve(nodes):
    if not nodes:
        return True

    node = nodes[0]

    for c in colors:
        if safe(node, c):
            color[node] = c

            if solve(nodes[1:]):
                return True

            del color[node]

    return False

if solve(list(graph.keys())):
    print("Map Coloring Solution:")
    for node, c in color.items():
        print(node, ":", c)
else:
    print("No solution exists")
```

```python
#Crypt-Arithmetic Puzzle

from itertools import permutations

letters = 'SENDMORY'

for p in permutations(range(10), len(letters)):
    d = dict(zip(letters, p))

    if d['S'] == 0 or d['M'] == 0:
        continue

    send = d['S']*1000 + d['E']*100 + d['N']*10 + d['D']
    more = d['M']*1000 + d['O']*100 + d['R']*10 + d['E']
    money = d['M']*10000 + d['O']*1000 + d['N']*100 + d['E']*10 + d['Y']

    if send + more == money:
        print("SEND =", send)
        print("MORE =", more)
        print("MONEY =", money)
        break
```

```python
#Local Search for CSP

import random

colors = ['Red', 'Green', 'Blue']

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

assignment = {v: random.choice(colors) for v in graph}

def conflicts(node, color):
    return sum(assignment[n] == color for n in graph[node])

for _ in range(1000):
    bad = [v for v in graph if conflicts(v, assignment[v]) > 0]

    if not bad:
        break

    node = random.choice(bad)
    assignment[node] = min(colors, key=lambda c: conflicts(node, c))

print("Map Coloring Solution:")
for node in assignment:
    print(node, ":", assignment[node])

if all(assignment[v] != assignment[n]
       for v in graph for n in graph[v]):
    print("All constraints satisfied")
else:
    print("Solution not found")
```
