```python
#Bayesian Reasoning – Medical Diagnosis
disease = 0.01
symptoms_given_disease = 0.90
symptoms_given_no_disease = 0.05

symptoms = (
    symptoms_given_disease * disease
    + symptoms_given_no_disease * (1 - disease)
)

result = (
    symptoms_given_disease * disease
) / symptoms

print("Probability of disease:", round(result * 100, 2), "%")
```

```python
#Fuzzy Inference System - To implement a Fuzzy Inference System (FIS) for controlling the speed of an electric vehicle based on traffic conditions.

def fuzzy_speed(traffic):
    if traffic <= 30:
        speed = 80
    elif traffic <= 60:
        speed = 50
    else:
        speed = 20

    return speed

traffic = int(input("Enter traffic level (0-100): "))

if 0 <= traffic <= 100:
    print("Traffic Level:", traffic)
    print("Recommended Speed:", fuzzy_speed(traffic), "km/h")
else:
    print("Invalid traffic level")
```
