```python
#simple reflex agent - thermostat

SET_TEMPERATURE = 25
current_temperature = int(input("Enter the current temperature: "))
if current_temperature < SET_TEMPERATURE:
    print("Heater ON")
else:
    print("Heater OFF")
```

```python
# Model-Based Agent - Autonomous Car

previous_obstacle = False
obstacle = input("Is there an obstacle ahead? (yes/no): ")
if obstacle.lower() == "yes":
    previous_obstacle = True
    print("Action: Stop the car")
else:
    if previous_obstacle:
        print("Action: Move slowly and check surroundings")
        previous_obstacle = False
    else:
        print("Action: Continue driving")
```

```python
# Goal-Based Agent - Warehouse Robot

goal = "Deliver Package"
location = input("Enter robot location (storage/destination): ")
package = input("Is the robot carrying a package? (yes/no): ")

if goal == "Deliver Package":
    if package.lower() == "no" and location == "storage":
        print("Action: Pick up the package")
    elif package.lower() == "yes" and location != "destination":
        print("Action: Move to destination")
    elif package.lower() == "yes" and location == "destination":
        print("Action: Deliver the package")
    else:
        print("Action: Search for the package")
```

```python
# Utility-Based Agent - Financial Trading Bot

buy_profit = int(input("Enter expected profit if BUY: "))
sell_profit = int(input("Enter expected profit if SELL: "))
hold_profit = int(input("Enter expected profit if HOLD: "))
if buy_profit >= sell_profit and buy_profit >= hold_profit:
    print("Action: BUY")
elif sell_profit >= buy_profit and sell_profit >= hold_profit:
    print("Action: SELL")
else:
    print("Action: HOLD")
```
