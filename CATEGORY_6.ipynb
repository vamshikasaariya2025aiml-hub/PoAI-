```python
#Min-Max Search Algorithm for Chess Game
def minimax(depth, node, maximizing, values):
    if depth == 2:
        return values[node]

    if maximizing:
        return max(
            minimax(depth + 1, node * 2, False, values),
            minimax(depth + 1, node * 2 + 1, False, values)
        )
    else:
        return min(
            minimax(depth + 1, node * 2, True, values),
            minimax(depth + 1, node * 2 + 1, True, values)
        )

values = [3, 5, 2, 9]

result = minimax(0, 0, True, values)

print("Best score:", result)
```

```python
#Alpha-Beta Pruning Algorithm for a Chess Game.
def alphabeta(depth, node, alpha, beta, maximizing, values):
    if depth == 2:
        return values[node]

    if maximizing:
        best = float('-inf')

        for i in range(2):
            score = alphabeta(depth + 1, node * 2 + i,
                              alpha, beta, False, values)
            best = max(best, score)
            alpha = max(alpha, best)

            if alpha >= beta:
                break

        return best

    else:
        best = float('inf')

        for i in range(2):
            score = alphabeta(depth + 1, node * 2 + i,
                              alpha, beta, True, values)
            best = min(best, score)
            beta = min(beta, best)

            if alpha >= beta:
                break

        return best


values = [3, 5, 2, 9]

result = alphabeta(0, 0, float('-inf'),
                   float('inf'), True, values)

print("Best score:", result)
```

```python
#Min-Max Search Algorithm for Tic-Tac-Toe Game

def check(board):
    wins = [(0,1,2), (3,4,5), (6,7,8),
            (0,3,6), (1,4,7), (2,5,8),
            (0,4,8), (2,4,6)]

    for a, b, c in wins:
        if board[a] == board[b] == board[c] != ' ':
            return board[a]
    return None

def minimax(board, maximizing):
    winner = check(board)

    if winner == 'X':
        return 1
    if winner == 'O':
        return -1
    if ' ' not in board:
        return 0

    if maximizing:
        best = -float('inf')
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'
                best = max(best, minimax(board, False))
                board[i] = ' '
        return best
    else:
        best = float('inf')
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'
                best = min(best, minimax(board, True))
                board[i] = ' '
        return best

board = [' '] * 9

for i in range(9):
    print(board[0:3])
    print(board[3:6])
    print(board[6:9])

    if check(board):
        print(check(board), "wins!")
        break

    if i % 2 == 0:
        best_score = -float('inf')
        move = -1

        for j in range(9):
            if board[j] == ' ':
                board[j] = 'X'
                score = minimax(board, False)
                board[j] = ' '

                if score > best_score:
                    best_score = score
                    move = j

        board[move] = 'X'
        print("Computer plays X")
    else:
        move = int(input("Enter position (0-8): "))
        if 0 <= move < 9 and board[move] == ' ':
            board[move] = 'O'
        else:
            print("Invalid position!")
            break

winner = check(board)
print("Final board:")
print(board[0:3])
print(board[3:6])
print(board[6:9])

if winner:
    print(winner, "wins!")
elif ' ' not in board:
    print("Draw!")
```
