````python
#Medical Diagnosis Using Knowledge Base

```python
facts = {
    "fever",
    "cough",
    "sore_throat"
}

rules = {
    "common_cold": ["cough", "sore_throat"],
    "flu": ["fever", "cough"],
    "throat_infection": ["fever", "sore_throat"]
}

def forward_chaining(facts, rules):
    new_facts = set(facts)
    changed = True

    while changed:
        changed = False

        for disease, symptoms in rules.items():
            if all(s in new_facts for s in symptoms):
                if disease not in new_facts:
                    new_facts.add(disease)
                    changed = True

    return new_facts

def backward_chaining(goal, facts, rules):
    if goal in facts:
        return True

    if goal not in rules:
        return False

    return all(
        backward_chaining(s, facts, rules)
        for s in rules[goal]
    )

print("Initial Facts:", facts)

result = forward_chaining(facts, rules)
print("Forward Chaining Results:", result)

goal = "flu"
if backward_chaining(goal, facts, rules):
    print("Backward Chaining: Flu may be indicated")
else:
    print("Backward Chaining: Flu not established")
````

````python
#Autonomous Vehicle

facts = {
    "red_light",
    "pedestrian_crossing"
}

rules = {
    "stop": ["red_light"],
    "slow_down": ["pedestrian_crossing"],
    "wait": ["stop"]
}

def forward_chaining(facts, rules):
    result = set(facts)
    changed = True

    while changed:
        changed = False

        for action, conditions in rules.items():
            if all(c in result for c in conditions):
                if action not in result:
                    result.add(action)
                    changed = True

    return result

def backward_chaining(goal, facts, rules):
    if goal in facts:
        return True

    if goal not in rules:
        return False

    return all(
        backward_chaining(c, facts, rules)
        for c in rules[goal]
    )

print("Initial Facts:", facts)

result = forward_chaining(facts, rules)
print("Forward Chaining:", result)

goal = "wait"

if backward_chaining(goal, facts, rules):
    print("Backward Chaining: Vehicle should wait")
else:
    print("Backward Chaining: Condition not established")
````
