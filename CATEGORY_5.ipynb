```python
#Greedy BFS – Tic-Tac-Toe

import heapq
start = ('X', 'O', 'X',
         'O', 'X', ' ',
         ' ', ' ', 'O')

def heuristic(state):
    return state.count(' ')

def greedy_bfs():
    pq = [(heuristic(start), start)]
    visited = set()

    while pq:
        h, state = heapq.heappop(pq)

        if state not in visited:
            visited.add(state)

            if state[6] == ' ':
                state = list(state)
                state[6] = 'X'
                print("Best Move:")
                print(state[0:3])
                print(state[3:6])
                print(state[6:9])
                return

            for i in range(9):
                if state[i] == ' ':
                    new_state = list(state)
                    new_state[i] = 'X'
                    heapq.heappush(
                        pq, (heuristic(tuple(new_state)),
                             tuple(new_state))
                    )

greedy_bfs()
```

```python
#A* Search - Finding the shortest path between cities

import heapq

graph = {
    'A': {'B': 2, 'C': 4},
    'B': {'D': 2, 'E': 5},
    'C': {'E': 2},
    'D': {'G': 3},
    'E': {'G': 3},
    'G': {}
}

h = {'A': 6, 'B': 4, 'C': 5,
     'D': 3, 'E': 2, 'G': 0}

def astar(start, goal):
    pq = [(h[start], 0, start, [start])]
    visited = {}

    while pq:
        f, g, node, path = heapq.heappop(pq)

        if node == goal:
            print("Path:", path)
            print("Cost:", g)
            return

        if node in visited and visited[node] <= g:
            continue

        visited[node] = g

        for nxt, cost in graph[node].items():
            new_g = g + cost
            new_f = new_g + h[nxt]
            heapq.heappush(
                pq, (new_f, new_g, nxt, path + [nxt])
            )

astar('A', 'G')
```

```python
#AO* Search - Medical Diagnosis System

graph = {
    'A': [[('B', 1), ('C', 1)], [('D', 4)]],
    'B': [],
    'C': [],
    'D': []
}

def ao_star(node):
    if not graph[node]:
        return 0

    best_cost = float('inf')
    best_path = []

    for children in graph[node]:
        cost = sum(edge + ao_star(child)
                   for child, edge in children)

        if cost < best_cost:
            best_cost = cost
            best_path = [child for child, edge in children]

    print(node, "->", best_path, "Cost:", best_cost)
    return best_cost

ao_star('A')
```

```python
#Memory Bounded Heuristic Search - Finding a Path in a Maze

graph = {
    'S': ['A'],
    'A': ['B', 'D'],
    'B': [],
    'D': ['G'],
    'G': []
}

h = {'S': 3, 'A': 2, 'B': 2, 'D': 1, 'G': 0}

def search(path, g, bound):
    node = path[-1]
    f = g + h[node]

    if f > bound:
        return f

    if node == 'G':
        return True

    minimum = float('inf')

    for nxt in graph[node]:
        if nxt not in path:
            path.append(nxt)
            result = search(path, g + 1, bound)

            if result is True:
                return True

            minimum = min(minimum, result)
            path.pop()

    return minimum

def ida_star():
    path = ['S']
    bound = h['S']

    while True:
        result = search(path, 0, bound)

        if result is True:
            print("Path:", " -> ".join(path))
            return

        if result == float('inf'):
            print("No solution")
            return

        bound = result

ida_star()
```
