```python
# Fully Observable Environment - Simple Chess Board
board = [
    ["r", "n", "b", "q", "k", "b", "n", "r"],
    ["p", "p", "p", "p", "p", "p", "p", "p"],
    [".", ".", ".", ".", ".", ".", ".", "."],
    [".", ".", ".", ".", ".", ".", ".", "."],
    [".", ".", ".", ".", ".", ".", ".", "."],
    [".", ".", ".", ".", ".", ".", ".", "."],
    ["P", "P", "P", "P", "P", "P", "P", "P"],
    ["R", "N", "B", "Q", "K", "B", "N", "R"]
]
def display_board():
    print("\nChess Board:")
    for row in board:
        print(" ".join(row))
    print()
display_board()
r1 = int(input("Enter source row (0-7): "))
c1 = int(input("Enter source column (0-7): "))
r2 = int(input("Enter destination row (0-7): "))
c2 = int(input("Enter destination column (0-7): "))
board[r2][c2] = board[r1][c1]
board[r1][c1] = "."
print("\nBoard after move:")
display_board()

# Partially Observable Environment - Driving in Fog

visibility = int(input("Enter visibility distance (in meters): "))
if visibility < 20:
    print("Action: Stop the car")
elif visibility < 50:
    print("Action: Drive slowly with fog lights ON")
else:
    print("Action: Drive at normal speed")
```

```python
# Known Environment - Simple Calculator

num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /): ")
num2 = float(input("Enter second number: "))
if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Division by zero is not allowed."
else:
    result = "Invalid operator."

print("Result:", result)

# Unknown Environment - Stock Market
import random
action = input("Choose an action (Buy/Sell/Hold): ")
market = random.choice(["Up", "Down"])
print("Market Movement:", market)
if action.lower() == "buy":
    if market == "Up":
        print("Result: Profit")
    else:
        print("Result: Loss")

elif action.lower() == "sell":
    if market == "Down":
        print("Result: Profit")
    else:
        print("Result: Loss")

elif action.lower() == "hold":
    print("Result: No transaction")

else:
    print("Invalid action")
```
