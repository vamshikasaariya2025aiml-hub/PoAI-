```python
            #Treasure Grid — Exploration vs Exploitation


#Exploitation: always choose the best neighboring cell.

#Exploration: randomly explore neighboring cells sometimes.
import random
grid = [
    [2, 4, 3, 1, 5],
    [6, 8, 2, 4, 3],
    [1, 5, 9, 2, 1],
    [3, 4, 6, 7, 2],
    [2, 1, 3, 5, 10]
]
rows = len(grid)
cols = len(grid[0])
def get_neighbors(r, c):
    neighbors = []
    if r > 0:
        neighbors.append((r - 1, c))

    if r < rows - 1:
        neighbors.append((r + 1, c))

    if c > 0:
        neighbors.append((r, c - 1))

    if c < cols - 1:
        neighbors.append((r, c + 1))

    return neighbors

# Exploitation
def exploitation():
    r, c = 0, 0
    path = [(r, c)]

    while True:
        neighbors = get_neighbors(r, c)

        best = max(neighbors, key=lambda x: grid[x[0]][x[1]])

        if grid[best[0]][best[1]] <= grid[r][c]:
            break

        r, c = best
        path.append((r, c))

    return path


# Exploration
def exploration():
    r, c = 0, 0
    path = [(r, c)]

    for i in range(8):
        neighbors = get_neighbors(r, c)

        # Sometimes explore randomly
        if random.random() < 0.4:
            next_cell = random.choice(neighbors)
        else:
            next_cell = max(
                neighbors,
                key=lambda x: grid[x[0]][x[1]]
            )

        r, c = next_cell
        path.append((r, c))

    return path

print("TREASURE GRID")
print()

for row in grid:
    print(row)

print("\nExploitation:")
path = exploitation()

for position in path:
    print(position, "Value =", grid[position[0]][position[1]])

print("\nExploration:")
path = exploration()

for position in path:
    print(position, "Value =", grid[position[0]][position[1]])

print("\nGlobal Optimum =", max(map(max, grid)))
```

```python
 #Eight Queens — Hill Climbing



  #This is the most important program for demonstrating local optimum.



  import random


# Count attacking pairs of queens
def calculate_conflicts(board):
    conflicts = 0

    for i in range(8):
        for j in range(i + 1, 8):

            # Same row
            if board[i] == board[j]:
                conflicts += 1

            # Same diagonal
            if abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts
    # Generate a neighboring state
def get_best_neighbor(board):
    best_board = board[:]
    best_conflicts = calculate_conflicts(board)

    for col in range(8):

        original_row = board[col]

        for row in range(8):

            if row == original_row:
                continue

            new_board = board[:]
            new_board[col] = row

            conflicts = calculate_conflicts(new_board)

            if conflicts < best_conflicts:
                best_conflicts = conflicts
                best_board = new_board

    return best_board, best_conflicts
    # Hill Climbing
def hill_climbing():

    board = [random.randint(0, 7) for _ in range(8)]

    print("Initial Board:")
    print(board)

    current_conflicts = calculate_conflicts(board)

    print("Initial Conflicts:", current_conflicts)

    while True:

        neighbor, neighbor_conflicts = get_best_neighbor(board)

        if neighbor_conflicts >= current_conflicts:
            break

        board = neighbor
        current_conflicts = neighbor_conflicts

        print("Board:", board)
        print("Conflicts:", current_conflicts)

        if current_conflicts == 0:
            break

    return board, current_conflicts
    board, conflicts = hill_climbing()

print("\nFinal Board:")
print(board)

print("Final Conflicts:", conflicts)

if conflicts == 0:
    print("Global Optimum Found!")
else:
    print("Local Optimum Reached!")
```

```python
 #Feature Selection — Simulated Annealing



  #For this experiment, we can use a small artificial dataset so that you don't need to download anything.

#The program will:

#Create data.
#Generate different feature combinations.
#Train a classifier.
#Calculate accuracy.
#Use Simulated Annealing to search for a better feature subset.



import random
import math

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score


# Create dataset
X, y = make_classification(
    n_samples=500,
    n_features=8,
    n_informative=5,
    n_redundant=1,
    random_state=42
)
# Calculate fitness
def fitness(state):

    selected = [i for i in range(8) if state[i] == 1]

    # At least one feature must be selected
    if len(selected) == 0:
        return 0

    X_selected = X[:, selected]

    X_train, X_test, y_train, y_test = train_test_split(
        X_selected,
        y,
        test_size=0.2,
        random_state=42
    )

    model = DecisionTreeClassifier(random_state=42)

    model.fit(X_train, y_train)

    prediction = model.predict(X_test)

    accuracy = accuracy_score(y_test, prediction)

    # Give a small reward for using fewer features
    score = accuracy - (0.01 * len(selected))

    return score
    # Create neighboring state
def get_neighbor(state):

    neighbor = state[:]

    position = random.randint(0, 7)

    if neighbor[position] == 0:
        neighbor[position] = 1
    else:
        neighbor[position] = 0

    return neighbor


# Simulated Annealing
def simulated_annealing():

    current = [random.randint(0, 1) for _ in range(8)]

    # Make sure at least one feature is selected
    if sum(current) == 0:
        current[0] = 1

    current_score = fitness(current)

    best = current[:]
    best_score = current_score

    temperature = 1.0
 for iteration in range(100):

        neighbor = get_neighbor(current)

        neighbor_score = fitness(neighbor)

        difference = neighbor_score - current_score

        # Accept better solution
        if difference > 0:

            current = neighbor
            current_score = neighbor_score

        else:

            # Sometimes accept worse solution
            probability = math.exp(difference / temperature)

            if random.random() < probability:
                current = neighbor
                current_score = neighbor_score

        # Update best solution
        if current_score > best_score:

            best = current[:]
            best_score = current_score

        # Reduce temperature
        temperature *= 0.95

    return best, best_score


best_state, best_score = simulated_annealing()

selected_features = [
    i + 1 for i in range(8)
    if best_state[i] == 1
]

print("SIMULATED ANNEALING")
print("-------------------")

print("Best Feature Selection:", best_state)

print("Selected Features:", selected_features)

print("Number of Features:", len(selected_features))

print("Best Score:", round(best_score, 4))
```
